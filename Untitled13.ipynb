```python
def unify(x, y, subs):
    if x == y:
        return subs

    if isinstance(x, str) and x.islower():
        subs[x] = y
        return subs

    if isinstance(y, str) and y.islower():
        subs[y] = x
        return subs

    return None


def unification(p1, p2):
    if p1[0] != p2[0]:
        return "Unification Failed"

    if len(p1) != len(p2):
        return "Unification Failed"

    subs = {}

    for i in range(1, len(p1)):
        result = unify(p1[i], p2[i], subs)

        if result is None:
            return "Unification Failed"

    return subs


p1 = ("Eats", "x", "Apple")
p2 = ("Eats", "Riya", "y")

result = unification(p1, p2)

if result == "Unification Failed":
    print(result)
else:
    print("Unification Successful")
    print("Substitution:", result)
```

```python
def unify(p1, p2):
    # Step 1: Check predicate symbols
    if p1[0] != p2[0]:
        return "Unification Failed"

    # Step 2: Check number of arguments
    if len(p1) != len(p2):
        return "Unification Failed"

    # Step 3: Create substitution dictionary
    subs = {}

    # Step 4: Compare arguments and apply substitutions
    for i in range(1, len(p1)):
        x = p1[i]
        y = p2[i]

        if x == y:
            continue

        elif x.islower():
            subs[x] = y

        elif y.islower():
            subs[y] = x

        else:
            return "Unification Failed"

    # Step 5: Return result
    return subs


# Main program
p1 = ("Eats", "x", "Apple")
p2 = ("Eats", "Riya", "y")

result = unify(p1, p2)

if result == "Unification Failed":
    print(result)
else:
    print("Unification Successful")
    print("Substitution:", result)
```
