```python

from itertools import product

def evaluate(expression, model):
    expression = expression.replace("<->", " == ")
    expression = expression.replace("->", " <= ")
    expression = expression.replace("&", " and ")
    expression = expression.replace("|", " or ")
    expression = expression.replace("~", " not ")

    return eval(expression, {"__builtins__": {}}, model)


def truth_table_entailment():
    print("Operators: & (AND), | (OR), ~ (NOT)")
    print("           -> (IMPLIES), <-> (BICONDITIONAL)")

    n = int(input("Enter number of propositions: "))
    symbols = input("Enter propositions separated by spaces: ").split()

    kb_count = int(input("Enter number of KB statements: "))
    kb_statements = []

    for i in range(kb_count):
        statement = input(f"Enter KB statement {i + 1}: ")
        kb_statements.append(statement)

    queries = ["Q", "P -> R", "R -> Q"]

    print("\nTruth Table:")

    for query in queries:
        print(f"\nChecking query: {query}")
        print(*symbols, "KB", "Query", sep="\t")

        entails = True
        found_model = False

        for values in product([True, False], repeat=n):
            model = dict(zip(symbols, values))

            kb_value = all(
                evaluate(statement, model)
                for statement in kb_statements
            )

            query_value = evaluate(query, model)

            print(*values, kb_value, query_value, sep="\t")

            if kb_value:
                found_model = True

                if not query_value:
                    entails = False

        if not found_model:
            print("KB is inconsistent.")
        elif entails:
            print(f"KB entails {query}: TRUE")
        else:
            print(f"KB entails {query}: FALSE")


truth_table_entailment()
```
