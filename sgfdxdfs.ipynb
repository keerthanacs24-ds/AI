```python

# BACKWARD CHAINING ALGORITHM

def clean(text):
    return text.lower().strip().replace(",", "").replace(".", "")


# Separate subject and predicate
def parse(sentence):
    words = clean(sentence).split()
    subject = words[0]
    predicate = " ".join(words[1:])
    return subject, predicate


# BACKWARD CHAINING
def backward_chaining(goal, facts, rules, visited=None):

    if visited is None:
        visited = set()

    # Check whether goal is already a fact
    if goal in facts:
        return True

    # Avoid repeated checking
    if goal in visited:
        return False

    visited = visited | {goal}

    # Check all rules
    for conditions, conclusion in rules:

        if conclusion == goal[1]:

            print("Checking:", goal[0], conclusion)

            result = True

            for condition in conditions:

                sub_goal = (goal[0], condition)

                print("Subgoal:", goal[0], condition)

                if not backward_chaining(
                    sub_goal, facts, rules, visited
                ):
                    result = False
                    break

            if result:
                return True

    return False


# INPUT FACTS
facts = []

n = int(input("Enter number of facts: "))

for i in range(n):
    fact = input("Enter fact: ")
    facts.append(parse(fact))


# INPUT RULES
rules = []

r = int(input("\nEnter number of rules: "))

for i in range(r):

    rule = clean(input("Enter rule: "))

    left, right = rule.split("then", 1)

    left = left.replace("if ", "", 1).strip()

    conditions = []

    for condition in left.split("and"):

        condition = condition.strip()

        if condition.startswith("x "):
            condition = condition[2:]

        conditions.append(condition)

    right = right.strip()

    if right.startswith("x "):
        right = right[2:]

    rules.append((conditions, right))


# INPUT GOAL
goal = parse(input("\nEnter goal: "))


# EXECUTION
if backward_chaining(goal, facts, rules):
    print("\nGOAL IS TRUE")
else:
    print("\nGOAL IS FALSE")
```

```python

# FORWARD CHAINING ALGORITHM

def forward_chaining(facts, rules, query):

    while True:
        new_facts = []

        for conditions, conclusion in rules:

            subjects = set(fact[0] for fact in facts)

            for subject in subjects:

                matched = True

                for condition in conditions:
                    if (subject, condition) not in facts:
                        matched = False
                        break

                if matched:
                    result = (subject, conclusion)

                    if result not in facts and result not in new_facts:
                        new_facts.append(result)

        if not new_facts:
            break

        facts.extend(new_facts)

        print("New facts inferred:")

        for fact in new_facts:
            print(fact[0], fact[1])

    return query in facts


# KNOWLEDGE BASE - FACTS
facts = [
    ("Tommy", "barks"),
    ("Tommy", "has four legs"),
    ("Tommy", "has sharp teeth")
]


# RULES
rules = [
    (["barks", "has four legs"], "is a dog"),

    (["has feathers", "lays eggs"], "is a bird"),

    (["is a dog"], "is a mammal"),

    (["is a mammal", "has sharp teeth"], "is a carnivore"),

    (["is a bird"], "can fly")
]


# QUERY
query = ("Tommy", "is a carnivore")


# EXECUTION
if forward_chaining(facts, rules, query):
    print("\nQUERY IS TRUE")
else:
    print("\nQUERY IS FALSE")
```
