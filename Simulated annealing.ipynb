```python

import random
import math

def cost(board):
    c = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                c += 1
    return c

def simulated_annealing():
    board = [random.randint(0, 7) for _ in range(8)]
    T = 100

    while T > 0.01:
        if cost(board) == 0:
            return board

        new = board[:]
        col = random.randint(0, 7)
        new[col] = random.randint(0, 7)

        delta = cost(new) - cost(board)

        if delta < 0 or random.random() < math.exp(-delta / T):
            board = new

        T *= 0.99

    return board if cost(board) == 0 else None

board = simulated_annealing()

if board:
    print("Solution:", board)
    print("Cost:", cost(board))
else:
    print("No solution found. Run again.")
```
