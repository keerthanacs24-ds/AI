```python
def cost(b):
    c = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if b[i] == b[j] or abs(b[i] - b[j]) == abs(i - j):
                c += 1
    return c

def show_board(b):
    for r in range(4):
        for col in range(4):
            print("Q" if b[col] == r else ".", end=" ")
        print()

def hill_climbing():
    b = [0, 0, 0, 0]

    print("Initial State:")
    show_board(b)
    print("Initial Cost:", cost(b))

    while cost(b) != 0:
        best = b[:]
        min_cost = cost(b)

        for i in range(4):
            for r in range(4):
                if r != b[i]:
                    n = b[:]
                    n[i] = r
                    if cost(n) < min_cost:
                        best = n
                        min_cost = cost(n)

        if best == b:
            print("Local minimum reached!")
            return

        b = best

    print("\nGoal State:")
    show_board(b)
    print("Goal Cost:", cost(b))

hill_climbing()
```
