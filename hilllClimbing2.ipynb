```python

import random
import math

# Calculate the number of attacking pairs of queens
def cost(state):
    attacks = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1

    return attacks


# Generate a random neighboring state
def get_neighbor(state):
    neighbor = state.copy()

    col1, col2 = random.sample(range(8), 2)
    neighbor[col1], neighbor[col2] = neighbor[col2], neighbor[col1]

    return neighbor


# Simulated Annealing algorithm
def simulated_annealing():
    current = random.sample(range(8), 8)
    T = 1000

    while T > 0.1:
        next_state = get_neighbor(current)

        delta_E = cost(current) - cost(next_state)

        if delta_E > 0:
            current = next_state
        else:
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                current = next_state

        if cost(current) == 0:
            return current

        T *= 0.99

    return current


# Display the solution
solution = simulated_annealing()

print("Solution:", solution)
print("Number of attacking pairs:", cost(solution))

if cost(solution) == 0:
    print("Solution found!")

    for row in range(8):
        print(" ".join("Q" if solution[col] == row else "."
                       for col in range(8)))
else:
    print("No solution found in this run. Run again.")
```
